```python
#Factorial of a Number

def Factorial(n):
    fact = 1
    og_n = n
    while (n!=0):
        fact = fact * n
        n-=1
    print(f"Factorial of {og_n}! is {fact}")

num = int(input("Enter a Number = "))
Factorial(num)
```

```python
#Excercise-1: Find the Average

def Average(t1,t2,t3):
    avg = (t1+t2+t3)/3
    print(f"The Average = {avg}")

test_1 = float(input("Test-1: "))
test_2 = float(input("Test-2: "))
test_3 = float(input("Test-3: "))

Average(test_1,test_2,test_3)
```

```python
#Problem-1: Simple Math Input/Output

#Obtaining & Storing the Values
num_1 = int(input("Number-1 = "))
num_2 = int(input("Number-2 = "))

#Caculation/Operation & Storing Values
sum = num_1 + num_2
product = num_1 * num_2

#Printing the Results
print(f"Sum of {num_1} & {num_2} is {sum}")
print(f"Product of {num_1} & {num_2} is {product}")
```

```python
#BMI Calculator

height = float(input("Height (m)= "))
weight = float(input("Weight (Kg)= "))

bmi = weight/(height**2)

print(f"Height: {height}m")
print(f"Weight: {weight}Kg")
print(f"Body-Mass-Index = {bmi:.2f} m/kg^2")
```
